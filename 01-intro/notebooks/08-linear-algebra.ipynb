## Multiplication

#### Vector * Vector

```python
def vetor_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0]
    result = 0.0
    n=u.shape[0]
    for i in range(n):
            result += u[i] * v[i]
    return result

vetor_vector_multiplication(u, v)
```

```python
# 1. We multiply 2 elements from u and v consecutivley 
# 2. We sum all the pair multiplications 
# (2 * 1) + (4 * 0) + (5 * 0) + (6 * 2) == 14  

# Method 1: using numpy dot product
print(u.dot(v))

#  Method 1:
print(u @ v)
```

#### Vector * Matrix

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])

v = np.array([1, 0, 0, 2])
```

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]
    num_rows =  U.shape[0]
    result = np.zeros(num_rows)
    for i in range(num_rows):
        result[i] = vetor_vector_multiplication(U[i], v)
    return result

matrix_vector_multiplication(U, v)
```

```python
# We perform vector * vector multiplication for each row of and the vector
# [(2 * 1) + (4 * 0) + (5 * 0) + (6 * 2)] = 14
# [(1 * 1) + (2 * 0) + (1 * 0) + (2 * 2)] = 5
# [(3 * 1) + (1 * 0) + (2 * 0) + (1 * 2)] = 5
# = [14, 5, 5]

U.dot(v)
```

#### Matrix * Matrix

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1], 
    [0, 2, 1],
    [2, 1, 0],
])

U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])
```

```python
# We perform the vector matrix multiplication of each row from V with the matrix U
# (1 * 2) + (0 * 4) + (0 * 5) + (2 * 6) = 5 
# (1 * 1) + (0 * 2) + (0 * 1) + (2 * 2) = 5  
# (1 * 3) + (0 * 1) + (0 * 2) + (2 * 1) = 5
# (1 * 2) + (0.5 * 4) + (2 * 5) + (1 * 6) = 20
# (1 * 1) + (0.5 * 2) + (2 * 1) + (1 * 2) = 6  
# (1 * 3) + (0.5 * 1) + (2 * 2) + (1 * 1) = 8.5  
# (2 * 2) + (1 * 4) + (1 * 5) + (0 * 6) = 13
# (2 * 1) + (1 * 2) + (1 * 1) + (0 * 2) = 5  
# (2 * 3) + (1 * 1) + (1 * 2) + (0 * 1) = 9
# = [ 14, 20, 13 ]
# = [ 	5,  6,  5 ]   
# = [ 	5, 8.5, 9 ]   
   

def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]
    num_rows = U.shape[0]
    num_cols = V.shape[1]
    
    result = np.zeros((num_rows, num_cols))
    for i in range(num_rows):
        result[:, i] = matrix_vector_multiplication(U, V[:, i])
    return result

matrix_matrix_multiplication(U, V)
```
