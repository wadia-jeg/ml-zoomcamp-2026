# Intruduction to Numpy

## Importing Numpy

```python
import numpy as np
```

## Creating arrays

```python
# From Zeros 
np.zeros(10)
```

```python
# From Ones
np.ones(10)
```

```python
# From a single value
np.full(10, 2.5)
```

```python
# From an array
a = np.array([1, 2, 3, 4])
a
```

```python
# From a range
# np.arrange(start, end)

np.arange(3, 10)
```

```python
# From a linspace
# np.linspace(start, end, number_of_elements)

np.linspace(0, 1, 11)
```

## Multi-dementional Arrays

```python
# np.zeros(shape=(rows, cols))
np.zeros((5, 2))
```

```python
n = np.array([
    [1, 2, 4],
    [4, 5, 6],
    [7, 8, 9],
]
)
n
```

## Randomly generated Arrays

```python
# Using Uniform Distribution
np.random.rand(5, 2)
```

```python
# For reproducebility 
# Fix a seed

np.random.seed(2)
np.random.rand(5, 2)
```

```python
# Using Normal Distribution

np.random.seed(2)
np.random.randn(5, 2)
```

```python
# Random Integer

np.random.randint(low=0, high=100, size=(5, 2))
```

## Indexing Numpy Arrays

```python
# One Dimentional array
a[2]
a
```

```python
# Assinging new value
a[2] = 4
a
```

```python
# One Dimentional array
# array[row, cols]

n[0, 1] = 20
n
```

```python
n[2] = [1, 1, 1]
n
```

```python
# Indexing cols
# ":"" means all elements

n[:, 2]
n
```
